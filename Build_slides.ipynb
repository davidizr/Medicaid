# Build ITS slides

Interactive front-end for `build_slides.py`. Every target produces **three slides**:

1. **Normalized Poisson ITS** — GLM Poisson with enrollment exposure offset (April 2022 interpolated).
2. **Normalized Negative Binomial ITS** — same design, NB2 likelihood.
3. **Quarterly Poisson-rate bar chart** with exact 95% CIs and a yearly summary table.

- **Individual codes** → each code gets its own 3 slides.
- **Aggregates** → each named group of codes is summed into one series that gets 3 slides.

Reads `claims_pivot.csv` (built by `Dataset_builder.ipynb`) and `Enrollees_Merged.xlsx`.

```python
%load_ext autoreload
%autoreload 2
import build_slides as bs
```

## 1. Individual codes

List the HCPCS codes you want as standalone slides. Each entry is either just the code

Set to `[]` if you only want aggregates.

```python
individual_codes = [
    ("33270"),
    ("33271"),
    ("33272"),
    ("93281"),
    ("93644")
]
```

## 2. Aggregates

Each key is a slide title; each value is the list of codes summed into one series.
Set to `{}` if you only want individual codes.

```python
aggregates = {
    # "Custom Aggregate 95921 and 95923": ["95921", "95923"],
    # "Cardiac monitoring": ["93241", "93242", "93243", "93244", "93245"],
}
```

## 3. Build the deck

Writes the `.pptx` and a companion utilization-rates `.csv`. The returned dict reports how
many targets rendered and lists anything skipped (e.g. a code missing from `claims_pivot.csv`).

```python
report = bs.build_deck(
    individual_codes=individual_codes,
    aggregates=aggregates,
    output_pptx="my_slides.pptx",
    utilization_rates_csv="my_slides_rates.csv",
    # intervention_month="2021-03-01",  # uncomment to override the default break point
)
report
```
